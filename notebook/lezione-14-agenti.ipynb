I passi dell'agente e i sei guardiani: le probabilità di passare indenni si moltiplicano.

Ogni numero calcolato qui viene confrontato con quello scritto nel libro: un segno ✓ vuol dire che coincidono.

```python
import math
import numpy as np

def controlla(nome, valore, atteso, tolleranza=0.01):
    """Confronta un valore calcolato con quello scritto nel libro."""
    ok = abs(valore - atteso) <= tolleranza
    segno = "✓" if ok else "✗ DIVERSO DAL LIBRO"
    print(f"{segno}  {nome}: calcolato {valore:.4g}, nel libro {atteso:g}")
    assert ok, nome

def conta(nome, valore, atteso):
    """Come controlla, per i numeri interi: devono coincidere esattamente."""
    ok = valore == atteso
    segno = "✓" if ok else "✗ DIVERSO DAL LIBRO"
    intero = lambda x: f"{x:,.0f}".replace(",", ".")
    print(f"{segno}  {nome}: calcolato {intero(valore)}, nel libro {intero(atteso)}")
    assert ok, nome
```

## I passi dell'agente

*Nel racconto: una ventina di passi, diciannove volte su venti; poco più di una volta su tre; la fila di secche.*

Se ogni passo riesce con probabilità p, indipendentemente dagli altri, n passi riescono tutti con probabilità pⁿ.

```python
controlla("diciannove volte su venti", 19 / 20, 0.95, 1e-9)
for p, n, atteso in [(0.95, 20, 0.36), (0.99, 20, 0.82), (0.9, 20, 0.12), (0.95, 5, 0.77)]:
    controlla(f"{n} passi al {p:.0%}", p**n, atteso, 0.005)

assert 1 / 3 < 0.95**20 < 0.4        # «poco più di una volta su tre»
assert 0.95**5 > 3 / 4               # «più di tre volte su quattro»
```

## Il formaggio svizzero

*Nel racconto: i sei guardiani; le fette con i buchi in posti diversi.*

Se ogni guardiano lascia passare per sbaglio una richiesta pericolosa con probabilità p, e i buchi sono indipendenti, sei guardiani la lasciano passare tutti con probabilità p⁶.

```python
p = 0.1
conta("sei guardiani con p = 0,1: una volta su", round(1 / p**6), 1_000_000)
print(f"ma se i buchi fossero tutti nello stesso posto, passerebbe ancora una volta su {1 / p:.0f}: guardiani diversi valgono solo se sbagliano in modi diversi")
```

## Prova tu

Quanti passi può fare un agente prima di scendere sotto il 50% di riuscita?

```python
for p in [0.9, 0.95, 0.99, 0.999]:
    n = math.floor(math.log(0.5) / math.log(p))
    print(f"con passi al {p:.1%}: al massimo {n} passi")
```
