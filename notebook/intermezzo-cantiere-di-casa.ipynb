# Intermezzo · Il cantiere di casa

Le quattro formule per capire se un modello ti sta in casa, e come andrà.

Ogni numero calcolato qui viene confrontato con quello scritto nel libro: un segno ✓ vuol dire che coincidono.

```python
import math
import numpy as np

def controlla(nome, valore, atteso, tolleranza=0.01):
    """Confronta un valore calcolato con quello scritto nel libro."""
    ok = abs(valore - atteso) <= tolleranza
    segno = "✓" if ok else "✗ DIVERSO DAL LIBRO"
    print(f"{segno}  {nome}: calcolato {valore:.4g}, nel libro {atteso:g}")
    assert ok, nome

def sigmoide(z):
    return 1 / (1 + math.exp(-z))
```

```python
def gb_tavole(parametri, bit):
    return parametri * bit / 8 / 1e9

def byte_per_token(strati, teste_kv, dim_testa, byte=2):
    return 2 * strati * teste_kv * dim_testa * byte

def token_al_secondo(banda_gb, tavole_gb, quaderno_gb):
    return banda_gb / (tavole_gb + quaderno_gb)

controlla("tavole del 27 miliardi a 4 bit (GB)", gb_tavole(27e9, 4), 13.5, 0)
bpt = byte_per_token(48, 8, 128)
controlla("byte per token", bpt, 196_608, 0)
controlla("quaderno a 32.000 token (GB)", bpt * 32_000 / 1e9, 6.3, 0.05)
controlla("quaderno a 128.000 token (GB)", bpt * 128_000 / 1e9, 25, 0.2)
for pagina, atteso in [(0, 27), (32_000, 19), (128_000, 10)]:
    controlla(f"token al secondo a {pagina} token", token_al_secondo(400, 15, bpt * pagina / 1e9), atteso, 0.5)
controlla("secondi alla prima parola su 10.000 token", 2 * 27e9 * 10_000 / 20e12, 27, 0.5)
```

## Kata per casa · Esercizio 1

```python
bpt = byte_per_token(32, 8, 128)
controlla("byte per token", bpt, 131_072, 0)
q16 = bpt * 32_000 / 1e9
controlla("quaderno a 16 bit (GB)", q16, 4.2, 0.05)
controlla("quaderno a 8 bit (GB)", q16 / 2, 2.1, 0.05)
controlla("token al secondo a pagina vuota", token_al_secondo(100, 4.5, 0), 22, 0.5)
controlla("con il quaderno a 16 bit", token_al_secondo(100, 4.5, q16), 11.5, 0.5)
controlla("con il quaderno a 8 bit", token_al_secondo(100, 4.5, q16 / 2), 15, 0.5)
```
